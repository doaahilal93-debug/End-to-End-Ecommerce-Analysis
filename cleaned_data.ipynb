```python
range= df.max(numeric_only= True ) - df.min(numeric_only= True ) 
print(range)
```

```python
Q1= df.quantile(0.25, numeric_only= True)
Q3= df.quantile(0.75, numeric_only= True)
```

```python
IQR= Q3 - Q1 
print(IQR)
```

```python
lower_bound= Q1-1.5*IQR
print(lower_bound)
```

```python
upper_bound= Q3+1.5*IQR
print(upper_bound)
```
